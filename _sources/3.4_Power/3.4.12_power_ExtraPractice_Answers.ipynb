## 1. Effect size for correlation

The researcher computes Pearson’s correlation coefficient between sleep duration (hours) and memory score and obtains:

$r$ = 0.22

### a) Interpreting effect size

- What does a correlation of $r = 0.22$ tell us about the *strength* of the relationship?
- Would you describe this as a small, medium, or large effect?
- What does this value of $r$ imply about the amount of overlap between high- and low-performing participants?

<span style="color:red">

- An observed correlation of r = 0.22 is a small-to-moderate positive relationship.
- Practically: r = 0.22 indicates a modest tendency for people who sleep more to score slightly higher, but a lot of variability remains (the scatter will be fairly diffuse).

</span>

## 2. Effect size $\neq$ statistical significance

The correlation is tested at the $\alpha = 0.05$ level using a one-tailed test (the researcher predicts that more sleep is associated with better memory).

### b) Sample size and significance

- Explain why a correlation of $r = 0.22$ might be statistically significant in a large sample, but not in a small sample.
- What does statistical significance tell us?


<span style="color:red">

- Statistical significance depends on both effect size and sample size.
- The same correlation (r = 0.22) can be:
  - Non-significant: in a small sample (because the estimate is noisy), and
  - Statistically significant: in a very large sample (because the standard error of r shrinks).

</span>

## 3. Power and sample size

The researcher now wants to consider whether the study was adequately powered.

### c) Post hoc power

- Using $r = 0.22$ as the effect size and $n = 60$, what is the power?
- Briefly explain why post hoc power analyses are **not** the ideal use of power analysis.

```python
#Convert r to t

r = 0.22
n = 60
t = (r*(n-2)**0.5)/((1-r**2)**0.5) # note **2 means 'squared', **0.5 means 'square root')
print(t)

#Convert t to d
d = t/(n**0.5)
print(d)
```

```python
# import required modules
from statsmodels.stats.power import TTestPower

analysis = TTestPower()

# solve for power
power = analysis.solve_power(effect_size=0.22, alpha=0.05, nobs=60, power=None, alternative='larger')
print('power = ' + str(power*100) + '%')
```

<span style="color:red">

- with n = 60 you have only roughly a 51% chance of detecting a ρ = 0.22 at α = 0.05.
- using the observed effect size from the study to compute power after data were collected is circular and often uninformative. Post-hoc power mainly restates the observed p-value in a different form; planning power before data collection is the preferred approach.
</span>

## 5. Effect size and variance

Now consider a *different* study.

A sports scientist compares reaction times (ms) between two groups:
- athletes who slept at least 8 hours;
- athletes who slept less than 6 hours.

The difference in mean reaction time between groups is **20 ms**.

### e) Effect size intuition

- Explain why a 20 ms difference could be a *large* effect in one study but a *small* effect in another.
- What role does within-group variability play in determining effect size?


<span style="color:red">

- The same difference in means (20 ms) can be:
  - Large if within-group standard deviations are small (e.g. sd ≈ 10 ms → Cohen’s d ≈ 2.0, very large), and
  - Tiny if within-group variability is large (e.g. sd ≈ 100 ms → Cohen’s d ≈ 0.2, small).
- Key point: effect size is the mean difference scaled by variability. Without knowing within-group variability, the raw difference (20 ms) tells you little about practical or statistical importance.
</span>

## 6. New Years Resolutions

A researcher investigates whether making a New Year’s resolution leads to an increase in exercise behaviour.

Participants report the average number of minutes they spend exercising per week at two time points:

- **December**, before the New Year
- **January**, after the New Year

Exercise time is measured in minutes per week, with higher values indicating more exercise.

Load the data below and determine if the study was properly powered

```python
resolution = pd.read_csv("https://raw.githubusercontent.com/jillxoreilly/StatsCourseBook_resources/main/data/new_years_resolution.csv")

resolution
```

```python
resolution.describe()
```

```python
results = stats.ttest_rel(resolution.ExerciseMinutes_January, resolution.ExerciseMinutes_December, alternative='greater')

d = results.statistic/(50**0.5)
print(d)
```

```python
# import required modules
from statsmodels.stats.power import TTestPower

# create an analysis of the correct form
analysis = TTestPower()

# solve for power
power = analysis.solve_power(effect_size=d, alpha=0.05, nobs=50, power=None, alternative='two-sided')
print('power = ' + str(power*100) + '%')
```
